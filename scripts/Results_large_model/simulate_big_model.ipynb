```python
import pandas as pd
import numpy as np
import maboss
```

```python
big_model_1 = maboss.loadBNet("/home/spankaew/Git/Git_Curie/AstroLogics/inferred_big_model/Signaling in Macrophage Activation_18433497/bn_0.bnet")
```

```python
big_model_1.param['time_tick'] = 1
big_model_1.param['max_time'] = 20
big_model_1.param['thread_count'] = 15
big_model_1.param['sample_count'] = 1000
big_model_1.network.set_output(big_model_1.network.names)

for node in big_model_1.network.names:
    big_model_1.network.set_istate(node, [0.5,0.5])
```

```python
sim = big_model_1.run()
sim_df = sim.get_nodes_probtraj()
```
