```python
import pandas as pd
import numpy as np
import maboss
import time
```

```python
big_model_1 = maboss.loadBNet("../../inferred_big_model/Signaling in Macrophage Activation_18433497/bn_0.bnet")
```

```python
big_model_1.param['time_tick'] = 1
big_model_1.param['max_time'] = 20
big_model_1.param['thread_count'] = 15
big_model_1.param['sample_count'] = 1000
big_model_1.network.set_output(big_model_1.network.names)

for node in big_model_1.network.names:
    big_model_1.network.set_istate(node, [0.5,0.5])
```

```python
start_time = time.time()
sim = big_model_1.run()
sim_df = sim.get_nodes_probtraj()
end_time = time.time()

elapsed_time = end_time - start_time
print(f"Simulation took {elapsed_time:.2f} seconds")
```

```python
import seaborn as sns

sns.clustermap(sim_df, row_cluster=False, figsize=(20,10))
```
