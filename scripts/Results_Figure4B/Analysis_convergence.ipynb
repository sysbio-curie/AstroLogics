# Invasion model : Model convergence analysis

```python
import pandas as pd 
import os
import sys
sys.path.append('../../src')
import astrologics as ast
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
os.chdir('../../')
model_path = 'models/Invasion/'
model = ast.ensemble(model_path, project_name = 'Invasion')
model.create_simulation()
```

```python
# Read network structure and set initial conditions
test = pd.read_csv(model_path + 'Invasion_0.bnet', sep = ',', header = None)
test[1] = 0.5  # Set all nodes to 0.5 probability initially
test_dict = dict(zip(test[0], test[1]))
```

```python
# Configure simulation parameters
model.simulation.update_parameters(max_time = 20,thread_count = 15, sample_count = 1000)
```

```python
convergence = model.simulation.analyze_sample_count_convergence(sample_counts=[50,100,250,500,750,
                                                                               1000,2500,5000,7500,10000],
                                                                               n_replicates = 5)
```

```python
std_table = convergence['summary']
std_table.to_csv('convergence_results.csv', index=False)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(9, 8))

metrics = [
    ('mean_std', 'Mean Std'),
    ('median_std', 'Median Std'),
    ('max_std', 'Max Std'),
    ('p95_std', 'P95 Std'),
]

for ax, (col, title) in zip(axes.flatten(), metrics):
    sns.lineplot(data=std_table, x='sample_count', y=col, marker='o', ax=ax)
    ax.set_title(title)
    ax.set_xlabel('Sample Count')
    ax.set_ylabel('Std')

plt.suptitle('Sample Count Convergence Analysis', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('convergence_analysis.png', dpi=500, bbox_inches='tight')
plt.savefig('convergence_results.pdf', bbox_inches='tight')
plt.show()
```
